### Library: NumPy
Core object: ndarray

A NumPy array is a fast, memory-efficient numerical array used mainly for mathematical and scientific computations.

✅ Key Characteristics

Homogeneous data (same data type)

Index is integer-based only

Optimized for fast numerical operations

N-dimensional (1D, 2D, 3D, etc.)

```python
import numpy as np

arr = np.array([10, 20, 30])
print(arr[0])
```

### Pandas Series

Library: pandas
Core object: Series

A Pandas Series is like a 1D labeled array.

You can think of it as:

NumPy array + labels (index)

✅ Key Characteristics

Has index labels

Can store mixed data types (though usually same type)

Built for data analysis

Works closely with DataFrame

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], index=['a', 'b', "tapojit", 100])
print(s['a'])
print(s["tapojit"])
print(s[100])
```
