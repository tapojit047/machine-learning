# Question 2: Library Inventory Management (Pandas)

## 1. DataFrame Creation

```python
import pandas as pd

data = {
    "Key": ["Book1", "Book2", "Book3", "Book4", "Book5"],
    "Title": [
        "The Great Gatsby",
        "To Kill a Mockingbird",
        "1984",
        "Pride and Prejudice",
        "The Catcher in the Rye"
    ],
    "Author": [
        "F. Scott Fitzgerald",
        "Harper Lee",
        "George Orwell",
        "Jane Austen",
        "J.D. Salinger"
    ],
    "Year": [1925, 1960, 1949, 1813, 1951],
    "Availability": [True, False, True, True, True]
}

library_df = pd.DataFrame(data)
library_df.set_index("Key", inplace=True)
library_df
```

## 2. Updating Data

```python
library_df.loc["Book2", "Availability"] = True

library_df
```

## 3. Search & Selection

```python
fitzgerald_books = library_df[library_df["Author"] == "F. Scott Fitzgerald"]

fitzgerald_books
```

```python
available_books = library_df[library_df["Availability"] == True]

available_books
```
